# Ingest bookings data
Copies the read-only sample data source into the project's dev schema,
verbatim, as the raw layer of the pipeline. Run manually; re-runnable.

Gather the data from `samples.wanderbricks.bookings` and copy it to `workspace.perceptron_dev.raw_bookings` 

```python
%sql
CREATE OR REPLACE TABLE
    workspace.perceptron_dev.raw_bookings
COMMENT 
    'Verbatim copy of samples.wanderbricks.bookings - raw layer, no transformation'
AS
    SELECT *
    FROM samples.wanderbricks.bookings;
```

Verify the copy

```python
%sql
SELECT 
    COUNT(*) AS rows_raw,
    COUNT(DISTINCT booking_id) AS distinct_bookings,
    MIN(created_at) AS earliest,
    MAX(created_at) AS latest
FROM 
    workspace.perceptron_dev.raw_bookings;
```
